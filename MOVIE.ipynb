```python
# import libraries
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
# loading datasets
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies.head(2)
```

```python
credits.head(2)
```

```python
# data preprocessing
movies = movies.merge(credits,on='title')
```

```python
movies.shape
```

```python
movies.info()
```

```python
## important columns for content-based prediction 
movies= movies[['movie_id','title', 'genres','keywords','overview','cast','crew']]
```

```python
movies.head(1)
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace = True)
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

```python
import ast
def convert(obj):
    l=[]
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
import ast
def convert1(obj):
    l=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter!=3:
            l.append(i['name'])
            counter+=1
        else:
             break
    return l
```

```python
movies['cast']=movies['cast'].apply(convert1)
```

```python
movies.head()
```

```python
import ast
def fetch_director(obj):
    l=[]
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply( lambda x: x.split())
```

```python
movies.head()
```

```python
movies['genres']=movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head()
```

```python
new_movies = movies[['movie_id','title','tags']]
```

```python
new_movies
```

```python
new_movies['tags']= new_movies['tags'].apply(lambda x:" ".join(x))
```

```python
new_movies.head()
```

```python
new_movies['tags']= new_movies['tags'].apply(lambda x:x.lower())
```

```python
new_movies.head()
```

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
 stemmer = PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(stemmer.stem(i))
    return " ".join(y)
```

```python
new_movies['tags']= new_movies['tags'].apply(stem)
```

```python
new_movies.head()
```

```python
## TEXT VECTORIZATION
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors= cv.fit_transform(new_movies['tags']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out(input_features=None)
```

```python
new_movies['tags'][0]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity[0].shape
```

```python
sorted(list(enumerate(map(float,similarity[0]))),reverse =True ,key = lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = new_movies[new_movies['title']== movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(map(float,distances))),reverse =True ,key = lambda x:x[1])[1:6]

    for i in movies_list:
            print(new_movies.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
